```python
# 1. 파일 읽기
with open('input.txt', 'r', encoding='utf-8') as file:
    content = file.read()

# 2. 공백이나 줄바꿈을 기준으로 단어들을 분리해서 리스트로 만들기
# .split()은 엔터, 스페이스, 탭 등을 전부 자동으로 잡아냅니다.
words = content.split()

# 3. 단어들을 콤마와 공백(, )으로 연결하여 하나의 문자열로 만들기
result = ", ".join(words)

# 4. 결과 출력 및 확인
print(result)

# (선택사항) 결과를 새 파일로 저장하고 싶다면 아래 주석을 해제하세요.
# with open('output.txt', 'w', encoding='utf-8') as out_file:
#     out_file.write(result)
```
